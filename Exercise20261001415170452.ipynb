問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

````python
year = int(input("請輸入年份："))

if year < 1912:
    ad_year = year + 1911
    roc_year = year
else:
    ad_year = year
    roc_year = year - 1911

animals = [
"monkey", "rooster", "dog", "pig",
"rat", "ox", "tiger", "rabbit",
"dragon", "snake", "horse", "sheep"
]

zodiac = animals[ad_year % 12]

print(f"西元 {ad_year} 年（民國 {roc_year} 年）的生肖是 {zodiac}")
````

1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。

使用「乘 2 取整數部分」的方法：
0.1 × 2 = 0.2 → 0

0.2 × 2 = 0.4 → 0

0.4 × 2 = 0.8 → 0

0.8 × 2 = 1.6 → 1

0.6 × 2 = 1.2 → 1

0.2 × 2 = 0.4 → 0
...
將取得的整數部分依序排列，可得：

0.1 = 0.00011001100110011...₂

也可以寫成：

0.1 = 0.0001(1001)₂

其中 (1001) 表示循環出現。

理由

在計算過程中，數值又回到：

0.2 → 0.4 → 0.8 → 1.6 → 0.6 → 1.2 → 0.2

因此相同的位元組合會不斷重複，形成無限循環小數。

看到的現象

0.1 無法用有限位數的二進位精確表示，而是：

0.00011001100110011...

這種無限循環形式。因此電腦只能儲存最接近 0.1 的近似值，而無法完全精確地儲存 0.1。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

由於 0.1 的二進位表示為：

0.00011001100110011...

是一個無限循環小數，因此無法用有限的位元完整儲存。

Python 的 float 採用 IEEE 754 雙精度浮點數（Double Precision）格式來儲存數值，共使用 64 個位元：

1 bit    Sign（符號）

11 bits  Exponent（指數）

52 bits  Fraction（尾數）

因為尾數位數有限，所以 Python 只能保留前面的有效數字，其餘部分必須進行捨入。

IEEE 754 規定預設使用「四捨六入五成雙（Round Half Even）」的方式進行捨入，也就是當數值剛好落在兩個可表示數值的正中間時，選擇較接近偶數的結果，以減少長期累積誤差。

因此 Python 並不是儲存精確的 0.1，而是儲存最接近 0.1 的 IEEE 754 浮點數值。這也是後面會出現浮點數誤差的原因。

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
## 原因
 
0.1 和 0.2 在二進位中都是無限循環小數，因此 Python 無法精確儲存它們，只能儲存最接近的近似值。
 
當兩個近似值相加時，就會產生微小誤差，因此結果顯示為：
 
```text
0.30000000000000004
```
 
而不是：
 
```text
0.3
```
 
## 解決方法
 
### 方法一：使用 round()
 
```python
print(round(0.1 + 0.2, 1))
```
 
輸出：
 
```text
0.3
```
 
### 方法二：使用 math.isclose()
 
```python
import math
 
print(math.isclose(0.1 + 0.2, 0.3))
```
 
輸出：
 
```text
True
```
 
### 方法三：使用 Decimal
 
```python
from decimal import Decimal
 
print(Decimal("0.1") + Decimal("0.2"))
```
 
輸出：
 
```text
0.3
```
 
## 結論
 
造成問題的原因是浮點數的二進位表示誤差。若只是顯示結果，可以使用 `round()`；若需要比較浮點數，建議使用 `math.isclose()`；若需要精確計算（例如金融計算），則應使用 `Decimal`。

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。

我認為不同領域應該使用不同的方法。
 
在科學計算中，浮點數本來就會有極小的誤差，因此通常使用 `float` 配合 `math.isclose()` 來判斷兩個數值是否接近。這種方法計算速度快，而且大部分科學與工程問題都能接受微小誤差。
 
例如：
 
```python
import math
 
math.isclose(0.1 + 0.2, 0.3)
```
 
在金融計算中，金額必須完全正確，即使是極小的誤差也可能導致結果錯誤。因此不建議使用 `float`，而應該使用 `Decimal` 進行計算。
 
例如：
 
```python
from decimal import Decimal
 
Decimal("0.1") + Decimal("0.2")
```
 
能夠得到精確的十進位結果。
 
另外，`round()` 雖然可以讓顯示結果變成 0.3，但它只是把誤差隱藏起來，並沒有真正解決浮點數誤差問題，因此較適合用在輸出結果的格式化，而不是精確計算。
 
## 結論
 
- 科學與工程計算：建議使用 `float` 搭配 `math.isclose()`
- 金融與會計計算：建議使用 `Decimal`
- `round()` 只適合顯示結果，不適合作為精確運算的方法
 
因此，沒有一種方法適用於所有情況，應根據實際需求選擇合適的方式。
