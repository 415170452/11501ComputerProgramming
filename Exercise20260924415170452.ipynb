## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？AI 模型中的權重通常使用浮點數（Floating Point Number）儲存，
最常見的是 FP32、FP16 和 BF16。
2. 什麼是 FP64？FP64（Double Precision Floating Point）是 64 位元浮點數。具有較高的計算精度，但需要較多記憶體，因此在大型 AI 模型中較少使用。
3. 什麼是 FP32？FP32（Single Precision Floating Point）是 32 位元浮點數。它在精度與運算效率之間取得平衡，因此長期以來是深度學習的標準格式。
4. 什麼是 FP16？FP16（Half Precision Floating Point）是 16 位元浮點數。與 FP32 相比可使用較少記憶體，
也能提高 GPU 運算速度，因此常用於深度學習訓練與推論。
5. 什麼是 BF16？BF16（Brain Floating Point 16）是一種 16 位元浮點格式。它保留與 FP32 類似的數值範圍，
同時降低記憶體需求，因此被廣泛應用於大型 AI 模型訓練。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為 AI 模型的權重通常是連續的實數，
需要進行大量的小數運算與梯度更新。int 只能表示整數，無法有效表示權重的微小變化。
此外，浮點數在 GPU 上具有更高的計算效率與較低的記憶體需求，
因此 AI 模型通常使用 FP32、FP16 或 BF16 儲存權重，而不是 Python 的 int。

1. Unlimited Precision 是什麼意思？Unlimited Precision（無限制精度）表示 Python 的 int 不受固定 32 位元或 64 位元限制，可以根據需要自動增加位數來儲存更大的整數。
2. 它是否代表可以儲存無限大的整數？不是。雖然理論上可以儲存非常大的整數，但仍受到電腦記憶體大小的限制，因此無法真正儲存無限大的數字。
3. CPython 如何實現這項特性？CPython 使用 PyLongObject 實作 int，並以動態配置記憶體的方式儲存數值。當整數變大時，系統會自動增加儲存空間，因此能表示遠大於 64 位元整數範圍的數值。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
Python 可以正常運作，是因為 int 採用 Arbitrary Precision（任意精度）設計。當數字變大時，CPython 會動態配置更多記憶體來儲存額外位數，因此能表示像 10**1000 這樣非常大的整數。然而，大型 AI 模型通常使用 FP32、FP16 或 BF16，而不是任意精度整數或高精度浮點數。主要原因如下：
 
1. 記憶體使用量
FP16 與 BF16 只需 16 位元，而 FP64 需要 64 位元。在具有數十億個參數的大型模型中，記憶體需求差異十分巨大。
 
2. 計算速度
較低精度的數值格式需要較少的運算資源，因此能提高訓練與推論速度。
 
3. 數值精度
雖然 FP16 與 BF16 的精度低於 FP32 或 FP64，但對大部分 AI 模型而言已足夠使用，模型效能通常不會受到明顯影響。
 
4. GPU 運算效率
現代 GPU 對 FP16 與 BF16 提供硬體加速與特殊最佳化，能大幅提升運算效率與吞吐量。
 
因此，AI 模型通常選擇 FP32、FP16 或 BF16，以在記憶體使用量、計算速度、數值精度與 GPU 效率之間取得較佳平衡。
