```python
total = 0

for i in range(1, 11):
    total += i

print(total)
```

```python
def summation(n):
    if n == 1:
        return 1
    return n + summation(n - 1)

print(summation(10))
```

```python
from functools import reduce

result = reduce(lambda x, y: x + y, range(1, 11))

print(result)
```
